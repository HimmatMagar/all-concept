```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 10))

sns.histplot(data=df, x='age', hue='insurance_premium_category', bins=20, kde=True, ax=ax[0, 0])
ax[0, 0].set_title("Distribution of age")

sns.histplot(data=df, x='weight', hue="insurance_premium_category", bins=20, kde=True, ax=ax[0, 1])
ax[0, 1].set_title("Distribution of weight")

sns.histplot(data=df, x='height', hue="insurance_premium_category", bins=20, kde=True, ax=ax[1, 0])
ax[1, 0].set_title("Distribution of height")

sns.histplot(data=df, x='income_lpa', hue="insurance_premium_category", bins=20, kde=True, ax=ax[1, 1])
ax[1, 1].set_title("Distribution of income")

plt.tight_layout()
plt.show()
```

```python
avg_age = df.groupby("insurance_premium_category")['age'].mean().reset_index()
avg_income = df.groupby("insurance_premium_category")['income_lpa'].mean().reset_index()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 10))

sns.barplot(data=avg_age, x="insurance_premium_category", y='age', ax=ax[0, 0])
ax[0, 0].set_title("Average Age")

sns.barplot(data=avg_income, x="insurance_premium_category", y='income_lpa', ax=ax[0, 1])
ax[1, 0].set_title("Average Income")

sns.countplot(data=df, x='smoker', ax=ax[1, 0])
ax[1, 0].set_title('Count plot of smoker')

sns.countplot(data=df, x='occupation', ax=ax[1, 1])
ax[1, 1].set_title('Count plot of occupation')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
